```python
from elasticsearch import Elasticsearch

es = Elasticsearch("http://localhost:9200")

doc = {
    "title": "Orange Juice",
    "content": "Orange juice is rich in vitamin C"
}

es.index(index="drinks", id=1, document=doc)

res = es.search(index="drinks", query={"match": {"content": "orange"}})
print(res)
```
