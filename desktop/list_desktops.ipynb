# List Desktops in Windows 10

*The following code shows All Desktop*

```python
import sys

try:
    from pyvda import VirtualDesktop, get_virtual_desktops
except ImportError:
    print("Error: pyvda is not installed. Run `pip install pyvda` first.")
    sys.exit(1)

def main():
    try:
        desktops = get_virtual_desktops()
        current = VirtualDesktop.current()
    except Exception as e:
        print(f"Failed to query virtual desktops: {e}")
        sys.exit(1)

    if not desktops:
        print("No virtual desktops found.")
        return

    print(f"Found {len(desktops)} virtual desktop(s):")
    for d in desktops:
        marker = "← Active" if d == current else ""
        # `d.number` is the 1-based index of the desktop
        print(f"  Desktop #{d.number:>2} {marker}")

if __name__ == "__main__":
    main()
```

### Show Active Virtual Desktop

```python
import sys
from pyvda import VirtualDesktop

def main():
    try:
        current = VirtualDesktop.current()
        # `.number` is 1-based index of the desktop
        print(f"Active Desktop: #{current.number}")
    except Exception as e:
        print(f"Error retrieving active desktop: {e}", file=sys.stderr)
        sys.exit(1)

if __name__ == "__main__":
    main()
```
